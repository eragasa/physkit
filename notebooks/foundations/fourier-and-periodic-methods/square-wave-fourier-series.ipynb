## Learning objectives

1. Evaluate a finite sum of odd sine harmonics through the package API.
2. Compare one-, five-, and twenty-term approximations.
3. Check odd symmetry and interpret behavior near jump discontinuities.


```python
import matplotlib.pyplot as plt
import numpy as np

from projectkoios.physkit.math.fourier_series import (
    OddSquareWaveFourierSeriesEvaluator,
)
```

```python
angles = np.linspace(-np.pi, np.pi, 1001, dtype=np.float64)
term_counts = (1, 5, 20)
approximations = {
    term_count: OddSquareWaveFourierSeriesEvaluator(term_count).execute(angles)
    for term_count in term_counts
}

figure, axis = plt.subplots(figsize=(9, 4.5))
for term_count, values in approximations.items():
    axis.plot(angles, values, label=f"{term_count} term(s)")
axis.set(
    xlabel="angle $x$ (radians)",
    ylabel="$S_M(x)$",
    title="Finite odd-harmonic square-wave series",
)
axis.grid(True)
axis.legend()
plt.show()
```

```python
probe_angles = np.array([0.2, 0.7, 1.4], dtype=np.float64)
evaluator = OddSquareWaveFourierSeriesEvaluator(term_count=20)

odd_symmetry_holds = np.allclose(
    evaluator.execute(-probe_angles),
    -evaluator.execute(probe_angles),
    rtol=0.0,
    atol=2e-15,
)
jump_values = evaluator.execute(
    np.array([0.0, np.pi], dtype=np.float64)
)
jump_midpoint_holds = np.allclose(
    jump_values,
    np.zeros(2, dtype=np.float64),
    rtol=0.0,
    atol=5e-15,
)

assert odd_symmetry_holds
assert jump_midpoint_holds
odd_symmetry_holds, jump_midpoint_holds
```

## Interpretation

Adding odd harmonics sharpens the plateaus away from the discontinuities. The finite series remains continuous, equals zero at the represented jump locations, and develops the familiar oscillatory overshoot near those jumps. These plots are illustrative numerical calculations, not evidence of pedagogical validation.
