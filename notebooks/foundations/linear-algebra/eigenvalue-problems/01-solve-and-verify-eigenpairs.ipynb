# Exercise 1 — Solve and verify eigenpairs

**Reading:** [Eigenvalue problems](../../../../docs/lecture-notes/foundations/linear-algebra/eigenvalue-problems/index.md)

For the real-symmetric matrix `A = [[2, -1], [-1, 2]]`, compute its eigenpairs and verify `A @ v = lambda * v` for every returned eigenvector.

```python
import numpy as np

from physkit.numerics.eigenproblems.real_symmetric import (
    RealSymmetricEigenpairSolver,
)
from physkit.units import MatrixQuantity, Unitless
```

```python
A = np.array([[2.0, -1.0], [-1.0, 2.0]])
result = RealSymmetricEigenpairSolver().execute(
    MatrixQuantity(A, Unitless())
)
eigenvalues = result.eigenvalues.magnitude
eigenvectors = result.eigenvectors.magnitude

print("eigenvalues:", eigenvalues)
print("eigenvectors (columns):\n", eigenvectors)
```

```python
residuals = A @ eigenvectors - eigenvectors * eigenvalues[np.newaxis, :]
residual_norms = np.linalg.norm(residuals, axis=0)
tolerance = 32.0 * np.finfo(np.float64).eps

print("residual norms:", residual_norms)
assert np.all(residual_norms < tolerance)
```

The residuals check the represented matrix equation. They do not establish the adequacy of any physical application.
