# Exercise 2 — Check an eigendecomposition

**Prerequisite:** Exercise 1

For the same real-symmetric matrix, verify that the eigenvectors are orthonormal and that `V @ diag(eigenvalues) @ V.T` reconstructs the matrix.

```python
import numpy as np

from physkit.numerics.eigenproblems.real_symmetric import (
    RealSymmetricEigenpairSolver,
)
from physkit.units import MatrixQuantity, Unitless
```

```python
A = np.array([[2.0, -1.0], [-1.0, 2.0]])
result = RealSymmetricEigenpairSolver().execute(
    MatrixQuantity(A, Unitless())
)
eigenvalues = result.eigenvalues.magnitude
V = result.eigenvectors.magnitude
```

```python
orthonormality_error = np.linalg.norm(V.T @ V - np.eye(2))
reconstruction = V @ np.diag(eigenvalues) @ V.T
reconstruction_error = np.linalg.norm(reconstruction - A)
tolerance = 32.0 * np.finfo(np.float64).eps

print(f"orthonormality error: {orthonormality_error:.3e}")
print(f"reconstruction error: {reconstruction_error:.3e}")
assert orthonormality_error < tolerance
assert reconstruction_error < tolerance
```

The two checks verify complementary properties: orthonormality of the basis and recovery of the represented matrix.
