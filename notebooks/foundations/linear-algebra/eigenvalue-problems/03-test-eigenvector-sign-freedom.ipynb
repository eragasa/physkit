# Exercise 3 — Test eigenvector sign freedom

**Prerequisite:** Exercises 1 and 2

Choose one computed eigenpair and verify that both `v` and `-v` satisfy the same eigenvalue equation. Then compare their projection matrices.

```python
import numpy as np

from physkit.numerics.eigenproblems.real_symmetric import (
    RealSymmetricEigenpairSolver,
)
from physkit.units import MatrixQuantity, Unitless
```

```python
A = np.array([[2.0, -1.0], [-1.0, 2.0]])
result = RealSymmetricEigenpairSolver().execute(
    MatrixQuantity(A, Unitless())
)
eigenvalue = result.eigenvalues.magnitude[0]
v = result.eigenvectors.magnitude[:, 0]
```

```python
positive_residual = np.linalg.norm(A @ v - eigenvalue * v)
negative_residual = np.linalg.norm(A @ (-v) - eigenvalue * (-v))
projector_difference = np.linalg.norm(np.outer(v, v) - np.outer(-v, -v))
tolerance = 32.0 * np.finfo(np.float64).eps

print(f"residual for v:          {positive_residual:.3e}")
print(f"residual for -v:         {negative_residual:.3e}")
print(f"projector difference:    {projector_difference:.3e}")
assert positive_residual < tolerance
assert negative_residual < tolerance
assert projector_difference < tolerance
```

The sign changes the vector representation but not its one-dimensional eigenspace.
