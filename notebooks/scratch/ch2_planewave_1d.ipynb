$$
    \psi(x,t) = e^{i(kx-\omega t)}
$$

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# define 1D grid
n_points = 500
length = 100
x = np.linspace(0, length, n_points)

k = 2 * np.pi * 3 / length
omega = 2 * np.pi
t = 0

psi =  np.exp(1j * (k * x - omega * t))

# plot
plt.plot(x, np.real(psi), label='Re[ψ(x)]')
plt.plot(x, np.imag(psi), label='Im[ψ(x)]', linestyle='--')
plt.title("1D Plane Wave")
plt.xlabel("x")
plt.ylabel("ψ")
plt.legend()
plt.grid(True)
plt.show()
```
