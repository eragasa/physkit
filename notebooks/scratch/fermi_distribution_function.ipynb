```python
import numpy as np

import numpy as np

class ConstantsSI:
    """
    Fundamental constants in SI units.
    """
    # Charge and masses
    q_e = 1.602176634e-19       # Elementary charge, C
    m_e = 9.1093837015e-31      # Electron rest mass, kg

    # Boltzmann and Planck constants
    k_B = 1.380649e-23          # Boltzmann constant, J/K
    h = 6.62607015e-34          # Planck constant, J*s
    hbar = h / (2 * np.pi)      # Reduced Planck constant, J*s

    # Derived units
    eV_to_J = q_e               # 1 eV in joules
    J_to_eV = 1 / q_e           # 1 J in eV


class ConstantsEV:
    """
    Same constants in electron-volt system for convenience.
    """
    k_B = 8.617333262e-5        # eV/K
    h = 4.135667696e-15         # eV*s
    hbar = 6.582119569e-16      # eV*s
    m_e = ConstantsSI.m_e       # keep SI for mass, usually needed


def get_fermi_dirac_distribution(
    E: float, 
    mu: float,
    T: float, 
    k_B: float = ConstantsEV.k_B
  ) -> float:
  """
  Compute the Fermi–Dirac distribution function.
  
  f(E) = 1 / (exp((E - mu) / (k_B T)) + 1)

  Args:
      E (float): Energy of the state (eV or J, must match units of mu and k_B*T).
      mu (float): Chemical potential / Fermi level (same units as E).
      k_B (float): Boltzmann constant (in same unit system as E and mu, e.g. eV/K).
      T (float): Absolute temperature (K).
  
  Returns:
      float: Occupation probability f(E), between 0 and 1.
  """
  # Avoid overflow when E >> mu at low T
  arg = (E - mu) / (k_B * T)
  # use numpy for stability
  return 1.0 / (np.exp(arg) + 1.0)
```
