```python
import numpy as np
import matplotlib.pyplot as plt

class LennardJonesPotential():

    # Lennard-Jones potential parameters
    epsilon = 1.0  # depth of the potential well
    sigma = 1.0    # distance at which the potential is zero

    @staticmethod
    def get_potential(r, epsilon, sigma):
        # Define the LJ potential
        V = 4 * epsilon * ((sigma / r)**12 - (sigma / r)**6)
        return V
    
    def get_force(r, epsilon, sigma):
        F = 4 * epsilon * (12 * sigma**12 / r**13 - 6 * sigma**6 / r**7)
        return F
     
# Generate r values, avoid r=0 to prevent singularity
r = np.linspace(0.8, 3.0, 500)
V = LennardJonesPotential.get_potential(
    r, 
    epsilon = 1.0, 
    sigma = 1.0
    )

# Plot
plt.figure(figsize=(6, 4))
plt.plot(r, V, label=r'$V_{\mathrm{LJ}}(r)$', color='blue')
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(2**(1/6)*sigma, color='red', linestyle='--', label=r'$r_0 = 2^{1/6}\sigma$')
plt.ylim(-1.2,1.2)
str_title = f"Lennard-Jones Potential, $\epsilon={epsilon:.2f}$, $\sigma={sigma:.2f}$"
plt.title(str_title)
plt.xlabel("Interatomic distance $r$")
plt.ylabel("Potential energy $V(r)$")
plt.legend()
plt.grid(False)
plt.tight_layout()
plt.show()
```

```python
# Compute the derivative of the Lennard-Jones potential (force)
def F_LJ(r, epsilon, sigma):
    return 4 * epsilon * (12 * sigma**12 / r**13 - 6 * sigma**6 / r**7)

F = F_LJ(r, epsilon, sigma)

plt.figure(figsize=(6, 4))
plt.plot(r, F, label=r'$F(r) = -\frac{dV}{dr}$', color='green')
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(2**(1/6)*sigma, color='red', linestyle='--', label=r'$r_0 = 2^{1/6}\sigma$')
plt.title("Force from Lennard-Jones Potential")
plt.xlabel("Interatomic distance $r$")
plt.ylabel("Force $F(r)$")
plt.ylim(-3,3)
plt.legend()
plt.tight_layout()
plt.show()
```
